# Connect to your Spark cluster

This notebook runs inside the `jupyter` container, which is pre-wired to the cluster.
Make sure the Spark profile is up (`docker compose --profile spark up -d`), then run the cells.

Keep the **Spark master UI** open at [http://localhost:8080](http://localhost:8080) — you'll see this
notebook appear as a running application, and every action you trigger becomes a job there.

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

# The ONLY configuration you'll ever need here: point at the cluster.
spark = (
    SparkSession.builder
    .master('spark://spark-master:7077')
    .appName('notebook')
    .getOrCreate()
)
spark
```

```python
# Sanity check: a distributed computation across your two workers.
df = spark.range(1_000_000).withColumn('x', F.rand())
print('partitions:', df.rdd.getNumPartitions())
df.agg(F.avg('x')).show()   # <- this is the ACTION that makes it all run
```

**Where files live:** this container sees your `workspace/` folder at both `~/work` and `/workspace`.
Always use `/workspace/...` paths when reading/writing with Spark — the workers mount the same
folder at that exact path, so driver and executors agree on where files are.
